## Gradient Descent

```python
data1=pd.read_table('P176.txt')
```

## Separate Dataset as Input X and Input Y

```python
X=data1['X']
Y=data1['Y']
X=np.array(X).reshape(-1,1)
Y=np.array(Y).reshape(-1,1)
```

## Scale the Data

```python
scalerx = MinMaxScaler()
X=scalerx.fit_transform(X)
scalery = MinMaxScaler()
Y=scalery.fit_transform(Y)
```

```python
def gradient_descent(X,Y,epochs=1000,lr=0.001):
    plt.scatter(X,Y)
    m=0
    c=0
    n=X.shape[0]
    for i in range(epochs):
        delta_m=(-2/n)*sum(X*(Y-m*X-c))
        delta_c=(-2/n)*sum(Y-m*X-c)
        m=m-lr*delta_m
        c=c-lr*delta_c
        pred=m*X+c
        plt.plot(X,pred)
    return m,c
```

```python
def predict(m,c,x):
    y=m*x+c
    return y
```

```python
m,c=gradient_descent(X,Y,epochs=10000,lr=0.001)
```

```python

print(m,c)
```

## Predict the response for X

```python
y_pred=predict(m,c,X)
```

## Print the Error

```python
print(mean_squared_error(scalery.inverse_transform(Y),scalery.inverse_transform(y_pred)))
```
